# Descriptive Analysis - Chile Dataset (EDA3)
This notebook performs a descriptive analysis on the EDA3 dataset, focusing on central tendency, dispersion, frequency distributions, and correlations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr
```

# Exercice 1
General overview

```python
# --- Import data ---
df = pd.read_csv('EDA3.csv', sep=r'\s+')

# --- Print information about data ---
print("--- Head ---")
print(df.head())

print("\n--- Shape ---")
print(df.shape)

print("\n--- Info ---")
print(df.info())

# This shows count, mean, std, min, max, percentiles, and median
print("\n--- Descriptive Statistics ---")
print(df.describe())
```

```python
# --- Count of each category in a categorical attributed series of values: value_counts() ---
# Mapping 'education' example to 'Sex' and 'vote' example to 'Party'
print("\nSex Count:")
print(df['Sex'].count())

print("\nValue counts for Sex:")
print(df['Sex'].value_counts())

print("\nValue counts for Party:")
print(df['Party'].value_counts())
```

```python
# --- To filter out the NA/NaN values in a specific column ---
# Note: In your specific dataset there are no NaNs, but this follows the requirement
strength_clean = df['Strength'].dropna()
print(f"\nStrength data after dropna (Size: {strength_clean.shape[0]})")
```

```python
# --- Statistics per group: groupby() ---
# Mapping 'region' example to 'Sex' and 'education/vote' to 'Sex/Party'
print("\nAverage Strength by Sex:")
print(df.groupby('Sex')['Strength'].mean())

print("\nGroupby Average Strength by Sex and Party:")
print(df.groupby(['Sex', 'Party'])['Strength'].mean())
```

*   **Data Types**: `ID`, `Age`, and `Strength` are integers, while `AgeGroup`, `Sex`, and `Party` are objects (categorical).
*   **Missing Values**: No missing values were found across any columns, indicating a clean dataset.

## Analyzing Strength by Age Group

In this section, we'll examine how 'Strength' varies across different 'AgeGroup' categories. This will involve looking at the distribution of age groups, calculating the average strength for each, and visualizing this relationship with a boxplot.

```python
# Value counts for AgeGroup
print("Distribution of AgeGroup:")
display(df['AgeGroup'].value_counts().sort_index().to_frame())

# Calculate average Strength for each AgeGroup
avg_strength_by_agegroup = df.groupby('AgeGroup')['Strength'].mean().to_frame()
print("\nAverage Strength by AgeGroup:")
display(avg_strength_by_agegroup)

# Box plot of Strength by AgeGroup
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='AgeGroup', y='Strength', palette='viridis', order=sorted(df['AgeGroup'].unique()))
plt.title('Strength Distribution by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Strength')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

# **Exercise 3**
Compute the covariance and correlation for two variables (age and strength,
for example). Show the relationship btw them.

```python
# Compute Covariance and Correlation
# Using Age and Strength as the two variables
print("Covariance Matrix:")
print(df[['Age', 'Strength']].cov())

print("\nCorrelation Matrix:")
# Use corr() from pandas.DataFrame
corr_matrix = df[['Age', 'Strength']].corr()
print(corr_matrix)
```

```python
# Create a Heatmap
# Use heatmap() from seaborn
plt.figure(figsize=(6, 4))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Heatmap of Age and Strength')
plt.show()
```

```python
# Pearson’s Correlation Coefficient
# Use pearsonr() SciPy function
corr_coeff, p_value = pearsonr(df['Age'], df['Strength'])
print(f"\nPearson correlation coefficient: {corr_coeff:.4f}")
print(f"P-value: {p_value:.4e}")
```

```python
# Scatter Plot
# Use scatter() from matplotlib.pyplot
plt.figure(figsize=(8, 6))
plt.scatter(df['Age'], df['Strength'], color='green', alpha=0.7)
plt.title('Relationship between Age and Strength')
plt.xlabel('Age')
plt.ylabel('Strength')
plt.grid(True)
plt.show()
```

### Summary for Exercise 3: Covariance and Correlation (Age vs. Strength)

This section explored the relationship between 'Age' and 'Strength' using covariance, correlation, and visualizations. Here are the key findings:

*   **Covariance Matrix**:
    *   The covariance between 'Age' and 'Strength' is approximately -85.41. A negative covariance suggests that as one variable increases, the other tends to decrease. The magnitude itself is hard to interpret directly.
*   **Correlation Matrix**:
    *   The correlation between 'Age' and 'Strength' is approximately -0.8679.
*   **Pearson's Correlation Coefficient**:
    *   **Value**: -0.8679. This indicates a **strong negative linear relationship** between 'Age' and 'Strength'. As age increases, strength tends to decrease significantly.
    *   **P-value**: 2.8621e-19 (which is a very small number, effectively 0). This extremely low p-value suggests that the observed strong negative correlation is highly unlikely to have occurred by random chance. Thus, the relationship is **statistically significant**.
*   **Heatmap**: Visually reinforces the strong negative correlation with a distinct color for the 'Age'-'Strength' cell.
*   **Scatter Plot**: The scatter plot clearly illustrates the strong negative trend. Data points generally show a downward slope from left to right, confirming that older individuals tend to have lower 'Strength' values.
